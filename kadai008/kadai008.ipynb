```python
#1 サンプルデータの取得
# Wine recognition dataset」を取得
from sklearn.datasets import load_wine

#インポートしたload_wine関数を実行し、datasetという変数に代入
dataset = load_wine()
dataset
```

```python
#2 データの構造と内容の確認
#ワインの成分を表示
dataset.data
```

```python
#要素の名称を確認。要素の名称はdatasetのfeature_namesキー
dataset.feature_names
```

```python
#Pandasをインポート
import pandas as pd
#DataFrame作成、変数dfに代入
#DataFrameのdataにワインの成分データを、columnsに要素名をセット

df = pd.DataFrame(data=dataset.data, columns=dataset.feature_names)
df.head()
```

```python
#DataFrameに、目的変数を追加
df['category'] = dataset.target
df.head()
```

```python
#2 サンプルデータの分割
#説明変数（ワインの成分）を変数Xに、目的変数（ワインの種別）を変数yに代入
X = dataset.data
y = dataset.target
```

```python
#model_selectionモジュールのtrain_test_split関数を実行
from sklearn.model_selection import train_test_split

#インポートしたtrain_test_split関数を実行。第1引数に説明変数を、第2引数に目的変数をセット
#学習データとテストデータは、7:3の比率で分割
#4つの要素をそれぞれ、変数X_train, X_test, y_train, y_testに代入
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=5)
```

```python
#実際に7:3に分割されているか、レコード数を確認
print(X.shape, X_train.shape, X_test.shape, y.shape, y_train.shape, y_test.shape)
```

```python
#3. 予測モデルのインスタンス化
#RandomForestClassifierクラスをインポート
from sklearn.ensemble import RandomForestClassifier

#インポートしたRandomForestClassifierクラスをインスタンス化して、modelという変数に代入
model = RandomForestClassifier(random_state=3)
```

```python
#4. 予測モデルの学習
#fitメソッドを実行。第1引数にX_trainを、第2引数にy_trainを指定。
model.fit(X_train, y_train)
```

```python
#5. 予測モデルの評価
#predictメソッドを利用することでy_pred(ワインの種別)を予測する
#predictメソッドの引数には、説明変数XのテストデータであるX_test(ワインの種別)を指定
y_pred = model.predict(X_test)
y_pred
```

```python
#目的変数y（ワインの種別）のテストデータを確認
y_test

#metricsモジュールのaccuracy_score関数をインポート
from sklearn.metrics import accuracy_score

#accuracy_score関数を実行
#第1引数にテストデータであるy_testを、第2引数に予測値であるy_predをセット
accuracy_score(y_test, y_pred)

```

```python
#6. 予測
import numpy as np

#predictメソッドを利用して成分データからワインの種別を予測
#引数には、予測したいワインの成分データを指定
X_real = np.array([
    [13, 1.6, 2.2, 16, 118, 2.6, 2.9, 0.21, 1.6, 5.8, 0.92, 3.2, 1011],
    [12, 2.8, 2.2, 18, 100, 2.5, 2.3, 0.25, 2.0, 2.2, 1.15, 3.3, 1000],
    [14, 4.1, 2.7, 24, 101, 1.6, 0.7, 0.53, 1.4, 9.4, 0.61, 1.6, 560]])

#予測したいワインの成分データであるX_realを引数に指定して、predictメソッドを実行
model.predict(X_real)
```
