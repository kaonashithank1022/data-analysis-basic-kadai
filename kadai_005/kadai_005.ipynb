```python
# Step1. matplotlib.pyplotメソッドとnumpyをインポートする
import matplotlib.pyplot as plt
import numpy as np

# Step2. Figure(描画領域全体)とAxes(ひとつひとつのプロットを描く領域)を作成する
fig, axes = plt.subplots(2, 3,tight_layout=True)

#S tep3. 各領域に、折れ線グラフ、sin関数、ヒストグラム、散布図のデータを表示する。 サブプロットの表示に対応したコードを書きます。

#折れ線グラフ
x1 = [0, 1, 2, 3]
y1 = [1, 2, -5, 2]
axes[0, 0].plot(x1, y1)

#sin関数
x2 = np.linspace(0, 10, 100)
y2 = 2 * np.sin(x2 * 2) + 2
axes[0, 1].plot(x2, y2)


#ヒストグラム
data3 = np.random.normal(20, 5, 1000)
axes[0, 2].hist(data3)

#散布図のデータ
np.random.seed(0)
x4 = np.random.rand(50) * 8
y4 = np.random.rand(50) * 8
axes[1, 0].scatter(x4, y4)

#下段中央 (1, 1) と 下段右 (1, 2) は空白のままでOK
# Step4. グラフを表示させる行「plt.show()」を書く


plt.show()



```
