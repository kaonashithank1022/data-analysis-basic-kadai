**どんな層かを一言で名づける視点**

各クラスターに「優良顧客」「新規のお客様」「しばらく来ていない層」のように、特徴ごとに分類すると整理しやすくなります。


**その層に何をしたいか、という販促の視点**

分類した層ごとに、「もっと利用してもらいたい」「引き止めたい」「育てたい」のどれに当てはまるかを考えると、打つべき施策のアイデアが浮かびやすくなります。
