**課題の内容**

NumPyを利用して、次の２つの行列の積を計算して結果を表示し、計算された行列の要素の最大値を求めて表示するコードを記述しましょう

作業上の注意事項
行列を、多次元配列の記法で書く際に、行と列を入れ違えない様にしましょう


**課題の作業手順**

Step1.
NumPyをインポートするコードを記述する

Step2.
AとBの行列を格納する多次元配列のコードを記述する

Step3.
行列の積の計算をするコードを記述する

Step4.
Step3.の計算結果を表示するコードを記述する

Step5.
行列の要素の最大値を求めるコードを記述する



```python
#Step1. NumPyをインポートするコードを記述する
import numpy as np

#Step2. AとBの行列を格納する多次元配列のコードを記述する
A = np.array([[0, 1], [2, 3],[4, 5]])
B = np.array([[0, 1, 2, 3], [4, 5, 6, 7]])

#Step3. 行列の積の計算をするコードを記述する
Product = np.dot(A, B)

#Step4. Step3.の計算結果を表示するコードを記述する
print(Product)

#Step5. 行列の要素の最大値を求めるコードを記述する
print(Product.max())
```
