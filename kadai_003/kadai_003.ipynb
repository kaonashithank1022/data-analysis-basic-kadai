```python
# Step1.NumPyをインポートするコードを記述する
import numpy as np

# Step2.AとBの行列を格納する多次元配列のコードを記述する
A = np.array([
    [0, 1],
    [2, 3],
    [4, 5]
])

B = np.array([
    [0, 1, 2, 3],
    [4, 5, 6, 7]
])

# Step3.行列の積の計算をするコードを記述する
C = np.dot(A, B)


# Step4.Step3.の計算結果を表示するコードを記述する
print("行列の積", C)


# Step5.行列の要素の最大値を求めるコードを記述する
max_value = C.max()  # np.max(C) でも可
print("最大値", max_value)
```
